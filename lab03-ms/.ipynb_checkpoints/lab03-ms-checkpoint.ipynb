# Part A — Iris: Lab 2's loop in one call

## A1. The altered training set of Lab 2

The cell below rebuilds the data of Lab 2, Part A: the 50-50 split with `random_state=10`, and the copy of the
training set with flowers 113 and 125 altered as in Lecture 9. Run it; do not change it.

```python
iris = load_iris(as_frame=True)
X = iris.data.copy()
X.columns = [c[:-5] for c in X.columns]
y = iris.target

X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=0.5, random_state=10)
X_train_alt, y_train_alt = X_train.copy(), y_train.copy()
y_train_alt.loc[125] = 0
X_train_alt.loc[113] = X_train_alt.loc[125]
y_train_alt.loc[113] = 1

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=SEED)   # the folds of Lab 2, step 4
print(X_train_alt.shape, y_train_alt.value_counts().sort_index().tolist())
```

## A2. `GridSearchCV`

In Lab 2, step 4, we wrote a loop: for each `max_depth` from 1 to 5, `cross_val_score`, then mean and standard deviation.
[`GridSearchCV`](https://scikit-learn.org/stable/modules/generated/sklearn.model_selection.GridSearchCV.html) does exactly
that loop. It is itself a model, with `fit`, `predict` and `score`; it needs three things:

- an **estimator** with the fixed hyperparameters: `DecisionTreeClassifier(criterion="entropy", random_state=10)`
- a **grid**: a dictionary `{hyperparameter name: list of values}`, here `{"max_depth": [1, 2, 3, 4, 5]}`
- the **folds**: `cv=cv`

1. Create the search object, call it `search_iris`, and `fit` it on the **altered training set** (never on the test set).
1. Print `best_params_` and `best_score_` (the best mean CV accuracy).
1. All the numbers are in the dictionary `cv_results_`. Turn it into a DataFrame and show the columns
   `param_max_depth`, `mean_test_score`, `std_test_score`, `rank_test_score`.

Careful with the names: in `cv_results_`, *test* means *the held-out fold of each round* — in Lecture 10a we called it
**validation**. The real test set never enters `GridSearchCV`.

```python
# 1. Create the search object and fit it on the altered training set
estimator = DecisionTreeClassifier(criterion="entropy", random_state=10)
param_grid = {"max_depth": [1, 2, 3, 4, 5]}

search_iris = GridSearchCV(estimator=estimator, param_grid=param_grid, cv=cv)
search_iris.fit(X_train_alt, y_train_alt)

# 2. Print best_params_ and best_score_
print("Best hyperparameters:", search_iris.best_params_)
print(f"Best mean CV accuracy: {search_iris.best_score_:.4f}")

# 3. Turn cv_results_ into a DataFrame and show the requested columns
cv_results_df = pd.DataFrame(search_iris.cv_results_)
columns_to_show = ["param_max_depth", "mean_test_score", "std_test_score", "rank_test_score"]

cv_results_df[columns_to_show]
```

**Answers.** *Write your answers here (double-click to edit).*

2. GridSearchCV chose a max_depth of 2 because its default tie-breaking behavior is to select the first parameter combination that achieves the highest score. In the parameter grid evaluated , the search evaluated depth 2 before depth 3, so depth 2 was locked in as the best parameter when it achieved the top score of 0.96. ## A3. The final model is already there

With the default `refit=True`, after the search `GridSearchCV` learns the best configuration **again on all the data it was given**
(the whole altered training set, not one of the folds), and stores it in `best_estimator_`. This is step 2 of the workflow of Lecture 10a.

1. How many trees did `fit` learn in total? Compute it from the size of the grid and the number of folds.
1. Print depth and number of leaves of `best_estimator_`.
1. Now, and only now, measure it **once** on the test set: `search_iris.score(X_test, y_test)` uses `best_estimator_`.

```python
# 1. Calculation of total trees fitted
grid_size = 5  # max_depth from 1 to 5
n_folds = 5    # 5-fold StratifiedKFold
total_trees = (grid_size * n_folds) + 1  # 25 CV trees + 1 final refit tree

print(f"Total trees trained: {total_trees}")

# 2. Print depth and number of leaves of best_estimator_
best_model = search_iris.best_estimator_
print(f"Best estimator max_depth: {best_model.get_depth()}")
print(f"Best estimator number of leaves: {best_model.get_n_leaves()}")

# 3. Measure accuracy on the test set
test_accuracy = search_iris.score(X_test, y_test)
print(f"Test set accuracy: {test_accuracy:.4f}")
```

**Answer.** *Total trees trained: 26, Cross-validation phase: 5 parameter configurations ($\text{max\_depth} \in \{1, 2, 3, 4, 5\}$) $\times$ 5 folds = 25 trees.Refit phase: Because refit=True by default, GridSearchCV fits 1 final model on the entire altered training set using the winning configuration ($\text{max\_depth}=2$).*

# Part B — The winner's curse, simulated

Lecture 10a, *Choosing on the test set: the winner's curse*: 20 "models" that answer at random, a test set of 50 records with two
balanced classes; we keep the model with the best test accuracy. Its true accuracy is 50%, but its test accuracy is about 63% on average.
Let us check it.

## B1. One random model

A model that flips a coin is right on each record with probability 0.5, independently: the number of correct answers on 50 records is a
**binomial** variable. `rng.binomial(n=50, p=0.5, size=...)` draws as many of them as we want.

1. Create a generator `rng = np.random.default_rng(0)`.
1. Draw the test accuracy of **one** random model on 50 records, 10 000 times (one value per repetition of the experiment).
   Print mean, minimum and maximum.

```python
# 1. Create the random number generator
rng = np.random.default_rng(0)

# 2. Draw accuracy (correct answers / 50) for 1 model across 10,000 repetitions
correct_counts = rng.binomial(n=50, p=0.5, size=10000)
accuracies = correct_counts / 50.0

# Print summary statistics
print(f"Mean accuracy: {accuracies.mean():.4f}")
print(f"Min accuracy:  {accuracies.min():.4f}")
print(f"Max accuracy:  {accuracies.max():.4f}")
```

## B2. Twenty random models, keep the best

Now each repetition has 20 models: draw a matrix with `size=(10_000, 20)` (one row per repetition, one column per model),
take the maximum of each row (`.max(axis=1)`), and average it. Draw the histogram of the winners' accuracies, with a vertical line
at the true accuracy 0.5.

```python
# B2. 20 random models — The Winner's Curse

# 1. Draw 20 models across 10,000 repetitions (shape: 10000 x 20)
models_20 = rng.binomial(n=50, p=0.5, size=(10000, 20)) / 50.0

# 2. Select the best (max) accuracy among the 20 models for each repetition
winning_accuracies = models_20.max(axis=1)

# 3. Print mean, minimum, and maximum of the winning model's accuracy
print(f"Mean best accuracy (Winner's Curse): {winning_accuracies.mean():.4f}")
print(f"Min best accuracy:                   {winning_accuracies.min():.4f}")
print(f"Max best accuracy:                   {winning_accuracies.max():.4f}")
```

## B3. More configurations, larger test sets

Write a function `winners_curse(n_test, n_models, reps=10_000)` returning the mean accuracy of the winner, and fill a table
with `n_test` in `[50, 900]` (rows) and `n_models` in `[1, 5, 20, 100]` (columns).

```python
def winners_curse(n_test, n_models, reps=10_000):
    rng = np.random.default_rng(0)
    draws = rng.binomial(n=n_test, p=0.5, size=(reps, n_models)) / n_test
    return draws.max(axis=1).mean()

n_test_list = [50, 900]
n_models_list = [1, 5, 20, 100]

# Generate the table of mean winning accuracies
table_data = {
    n_m: [winners_curse(n_t, n_m) for n_t in n_test_list]
    for n_m in n_models_list
}

df_winners_curse = pd.DataFrame(table_data, index=n_test_list)
df_winners_curse.index.name = "n_test \\ n_models"
df_winners_curse
```

**Questions.**

1. How does the optimism of the winner change with the number of configurations, and with the size of the test set?
1. The NordicMart grid of Part C has 30 configurations. Why must they compete on validation data, and not on the 900 test customers?

**Answers.** *

1. With the number of configurations ($n_{\text{models}}$): Optimism increases. The more configurations you evaluate, the higher the chance that at least one model achieves a high score purely due to random fluctuation/luck.With the size of the test set ($n_{\text{test}}$): Optimism decreases. As the sample size grows, variance shrinks ($\sigma \propto \frac{1}{\sqrt{n_{\text{test}}}}$) and sample accuracy concentrates around the true probability ($0.50$), making extreme lucky runs far less likely.
2. If 30 configurations are evaluated on the 900 test customers to pick the best one, the test set is no longer an independent benchmark—it is actively used for tuning.
The winning configuration's score on those 900 customers would suffer from the winner's curse (optimistic bias) and fail to reflect true performance on unseen future data.
By tuning hyperparameters on validation data (or cross-validation), the 900 test customers remain completely untouched until the end, providing an honest, unbiased estimate of the final model's performance.*

## B4. The winner on fresh data

In B2 we knew the truth (0.50) because we built the models. In practice we do not: what reveals the curse is
**measuring the winner again on data that played no part in choosing it**.

For each of the 10 000 repetitions of B2, take the winner and score it on **50 new records**. Since every model flips
a coin, the winner too is right with probability 0.5 on each new record: draw its fresh accuracy with `rng.binomial`.

1. Print the mean accuracy of the winners on the selection set (B2) and on the fresh set.
1. Draw the two histograms on the same axes (use `alpha=0.6` to see both).

```python
# Set up random generator matching initial state
rng = np.random.default_rng(0)

# Re-run B2 selection step: 20 models over 50 test records across 10,000 repetitions
models_20 = rng.binomial(n=50, p=0.5, size=(10000, 20)) / 50.0
winning_accuracies = models_20.max(axis=1)

# B4: Evaluate the winner of each repetition on 50 new independent records
fresh_accuracies = rng.binomial(n=50, p=0.5, size=10000) / 50.0

# 1. Print mean accuracies
print(f"Mean accuracy on selection set: {winning_accuracies.mean():.4f}")
print(f"Mean accuracy on fresh test set: {fresh_accuracies.mean():.4f}")

# 2. Draw histograms on the same axes
plt.hist(winning_accuracies, bins=20, alpha=0.6, label="Selection set")
plt.hist(fresh_accuracies, bins=20, alpha=0.6, label="Fresh test set")
plt.xlabel("Accuracy")
plt.ylabel("Frequency")
plt.title("Winner's Curse: Selection Set vs. Fresh Data")
plt.legend()
plt.show()
```

**Question.** The winner did not change between the two measures. Why did its accuracy drop?

**Answer.** *The accuracy dropped because the high score on the selection set was driven by favorable random sampling noise (luck) rather than genuine predictive capability.Selection Bias: When evaluating 20 random models on the selection set, the "winner" is simply the model that experienced the most optimistic random variations on those specific 50 records.Independence of New Data: The fresh set consists of 50 entirely new, independent coin flips. The lucky random errors that boosted the winner's score on the first dataset do not carry over to the new dataset.Regression to the Mean: Because the model's true underlying success probability is fixed at $p = 0.50$, its performance on fresh, unbiased data naturally reverts back to its true expected mean of 50%.*

### The winner's curse, in one sentence

> When you pick the best of many noisy scores, the winning score is biased upwards: the winner was chosen partly for its luck,
> and luck does not repeat on new data.

The name comes from **auctions** (Capen, Clapp and Campbell, 1971, bids for offshore oil leases): each company estimates the value of
the field with some error, and the one that wins is the one that overestimated it most — so the winner tends to overpay.
Our configurations "bid" with their validation scores, and the highest bid is the most optimistic one.

Another picture: give a 10-question true/false quiz to 100 students who all answer at random. The best one scores about 9/10.
Give that student a new quiz: about 5/10.

### Skill and luck ★

Real models are not all equal. Give the 20 models true accuracies evenly spread from 0.50 to 0.60
(`np.linspace(0.50, 0.60, 20)`); on 50 records, pick the winner, then score it on 50 fresh records. Over 10 000 repetitions:
mean selection score of the winner, mean fresh score, mean **true** accuracy of the winner, and how often the winner is really the best model (0.60).

```python
rng = np.random.default_rng(0)
reps = 10_000
n_test = 50

# 20 models with true accuracies evenly spaced from 0.50 to 0.60
true_p = np.linspace(0.50, 0.60, 20)

# 1. Simulate selection set performance across 10,000 repetitions
selection_counts = rng.binomial(n=n_test, p=true_p, size=(reps, 20))
selection_acc = selection_counts / n_test

# 2. Pick the winner (first occurrence of max score) for each repetition
winning_idx = selection_acc.argmax(axis=1)

winner_selection_acc = selection_acc[np.arange(reps), winning_idx]
winner_true_acc = true_p[winning_idx]

# 3. Score the winner of each repetition on 50 fresh records
fresh_counts = rng.binomial(n=n_test, p=winner_true_acc)
winner_fresh_acc = fresh_counts / n_test

# 4. Calculate required statistics
mean_selection = winner_selection_acc.mean()
mean_fresh = winner_fresh_acc.mean()
mean_true = winner_true_acc.mean()
pct_best = (winning_idx == 19).mean() * 100  # Index 19 corresponds to true accuracy 0.60

print(f"Mean selection score of winner: {mean_selection:.4f}")
print(f"Mean fresh score of winner:     {mean_fresh:.4f}")
print(f"Mean true accuracy of winner:    {mean_true:.4f}")
print(f"How often winner is best model:  {pct_best:.2f}%")
```

# Part C — NordicMart: model selection for the churn tree

## C1. Data and the locked test set

Same preparation as Lab 2, Part B:

1. Load the file into `df`.
1. Features `FEATURES = ["tier", "age", "days_since_last_buy", "annual_spend", "web_sessions_month", "tickets_90d"]`
   in a copy `Xn`, with `tier` coded Basic = 0, Silver = 1, Gold = 2; missing values stay `NaN`. Target `yn = df["churned"]`.
1. Stratified 70/30 split with `random_state=SEED`: `Xn_train, Xn_test, yn_train, yn_test`.
1. Print the sizes and the number of churners of the two parts, and the baseline accuracy ("nobody churns") on the training part.

From here to section C8 the test set is **locked**.

```python
# 1. Load the dataset
df = pd.read_csv(NORDICMART_FILE)

# 2 & 3. Prepare features and map 'tier' values to integers
FEATURES = ["tier", "age", "days_since_last_buy", "annual_spend", "web_sessions_month", "tickets_90d"]
tier_map = {"Basic": 0, "Silver": 1, "Gold": 2}

Xn = df[FEATURES].copy()
Xn["tier"] = Xn["tier"].map(tier_map)
yn = df["churned"]

# 4. Stratified 70/30 train/test split
Xn_train, Xn_test, yn_train, yn_test = train_test_split(
    Xn, yn, test_size=0.3, stratify=yn, random_state=SEED
)

# 5. Print sizes, churner counts, and baseline accuracy
print(f"Train set size: {Xn_train.shape[0]} | Churners: {yn_train.sum()}")
print(f"Test set size:  {Xn_test.shape[0]} | Churners: {yn_test.sum()}")

baseline_acc = (yn_train == 0).mean()
print(f"Baseline accuracy ('nobody churns') on train: {baseline_acc:.4f}")
```

## C2. The grid

The example of Lecture 10a, *Searching the hyperparameter space*:

| hyperparameter | values |
|---|---|
| `max_depth` | 2, 3, 4, 5, 6 |
| `min_samples_leaf` | 1, 10, 30 |
| `criterion` | `"gini"`, `"entropy"` |

1. Write it as a dictionary `param_grid`.
1. `ParameterGrid(param_grid)` is the list of all the combinations, each one a dictionary. How many are there? Print the first two.

```python
# 1. Define the dictionary of hyperparameter values
param_grid = {
    "max_depth": [2, 3, 4, 5, 6],
    "min_samples_leaf": [1, 10, 30],
    "criterion": ["gini", "entropy"]
}

# 2. Convert to ParameterGrid and inspect combinations
grid = ParameterGrid(param_grid)

print(f"Total configurations: {len(grid)}")
print("\nFirst two combinations:")
for params in list(grid)[:2]:
    print(params)
```

A configuration dictionary can be passed to the constructor with `**`: `DecisionTreeClassifier(random_state=0, **config)`
is the same as writing `DecisionTreeClassifier(random_state=0, criterion="gini", max_depth=2, min_samples_leaf=1)`.

## C3. Recipe 3: train / validation / test

1. Split the **training part** again, stratified: 600 customers for validation, the rest for training, `random_state=SEED`.
   Call the parts `X_tr, X_val, y_tr, y_val`. How many churners are in the validation set?
1. For each of the 30 configurations: learn a tree (`random_state=0`) on `X_tr`, compute its accuracy on `X_val`.
   Collect configuration and validation accuracy in a DataFrame `val_results`, sorted from best to worst; show the top 8.

```python
# 1 & 2. Split Xn_train into training (X_tr, y_tr) and validation (X_val, y_val) sets
X_tr, X_val, y_tr, y_val = train_test_split(
    Xn_train, yn_train, test_size=600, stratify=yn_train, random_state=SEED
)

print(f"Churners in validation set: {y_val.sum()} out of {len(y_val)} customers")

# 3 & 4. Evaluate each of the 30 configurations on X_tr and score on X_val
results = []

for params in grid:
    tree = DecisionTreeClassifier(**params, random_state=0)
    tree.fit(X_tr, y_tr)
    val_acc = tree.score(X_val, y_val)

    # Store hyperparameter dictionary with validation accuracy
    res_entry = dict(params)
    res_entry["val_accuracy"] = val_acc
    results.append(res_entry)

# Build DataFrame sorted from best to worst validation accuracy
val_results = (
    pd.DataFrame(results)
    .sort_values(by="val_accuracy", ascending=False)
    .reset_index(drop=True)
)

val_results.head(8)
```

## C4. Recipe 4: cross-validation with `GridSearchCV`

1. Create `cv_n = StratifiedKFold(n_splits=5, shuffle=True, random_state=0)` (the folds of Lab 2 and Lecture 10a).
1. Create `search = GridSearchCV(...)` with the estimator `DecisionTreeClassifier(random_state=0)`, the grid `param_grid`,
   `cv=cv_n` and `scoring="accuracy"`. Accuracy is the default for a classifier: we write it explicitly because on Wednesday
   this is the argument we will change.
1. `fit` it on the training part. How many trees are learned?
1. Print `best_params_` and `best_score_`.

```python
# 1. Create 5-fold Stratified K-Fold generator
cv_n = StratifiedKFold(n_splits=5, shuffle=True, random_state=0)

# 2. Instantiate GridSearchCV
estimator = DecisionTreeClassifier(random_state=0)
search = GridSearchCV(
    estimator=estimator,
    param_grid=param_grid,
    cv=cv_n,
    scoring="accuracy"
)

#3. Fit on the full training set (Xn_train, yn_train)
search.fit(Xn_train, yn_train)

# 4. Print winning parameters and best score
print("Best hyperparameters:", search.best_params_)
print(f"Best mean CV accuracy: {search.best_score_:.4f}")
```

```python
fig, ax = plt.subplots(figsize=(10, 6))

# Baseline accuracy ("nobody churns") on training set
baseline_acc = (yn_train == 0).mean()

# Group by (criterion, min_samples_leaf) to get 6 lines
groups = res.groupby(["criterion", "min_samples_leaf"])
offsets = np.linspace(-0.18, 0.18, len(groups))

for (criterion, leaf), group in groups:
    # Sort group by max_depth
    group_sorted = group.sort_values("max_depth")
    
    # Pick next offset
    offset = offsets[len(ax.lines) // 2] if len(ax.lines) > 0 else offsets[0]
    
    ax.errorbar(
        x=group_sorted["max_depth"] + offset,
        y=group_sorted["mean_test_score"],
        yerr=group_sorted["std_test_score"],
        label=f"{criterion}, leaf={leaf}",
        marker="o",
        capsize=3,
        alpha=0.85
    )

# Draw baseline line
ax.axhline(baseline_acc, color="black", linestyle=":", label=f"Baseline ({baseline_acc:.4f})")

ax.set_xlabel("max_depth")
ax.set_ylabel("Mean CV Accuracy")
ax.set_xticks([2, 3, 4, 5, 6])
ax.set_title("GridSearch Accuracy vs. max_depth (±1 SD)")
ax.legend(bbox_to_anchor=(1.05, 1), loc="upper left")
ax.grid(True, linestyle=":", alpha=0.6)

plt.tight_layout()
plt.show()
```

## C5. Within the noise, take the simplest

Lecture 10a, *Reading the cross-validation results*: among the configurations whose mean is within **one standard deviation** of the best,
take the **simplest**. For a tree, simpler means: smaller `max_depth` first; for the same depth, larger `min_samples_leaf`
(more customers per leaf, fewer leaves).

This is a **heuristic**, not a statistical test. The standard deviation of the 5 fold scores is not a confidence interval:
the rounds share most of their training data, so they are not independent, and 5 values give only a rough estimate.
We use it as a scale for the noise when comparing configurations scored on the same folds — the same scale as the error bars of C4.
(Hastie, Tibshirani and Friedman's *one-standard-error rule* uses std/√k instead, about 2.2 times narrower for k = 5: our version is more conservative.)

1. Compute the threshold: mean of the best configuration minus its standard deviation.
1. Select the candidates of `res` at or above the threshold. How many are there?
1. Sort the candidates by `max_depth` (increasing), then `min_samples_leaf` (decreasing), then `mean_test_score` (decreasing),
   and take the first one: call its hyperparameters `chosen` (a dictionary, with integer values for the two numbers).

```python
# 1. Compute the 1-SD threshold
best_mean = res.iloc[0]["mean_test_score"]
best_std = res.iloc[0]["std_test_score"]
threshold = best_mean - best_std

print(f"Best mean CV accuracy: {best_mean:.4f}")
print(f"Best std dev:          {best_std:.4f}")
print(f"1-SD Threshold:        {threshold:.4f}")

# 2. Select candidates at or above the threshold
candidates = res[res["mean_test_score"] >= threshold].copy()
print(f"\nNumber of candidate configurations within 1 SD: {len(candidates)}")

# 3. Sort candidates: max_depth (increasing), min_samples_leaf (decreasing), mean_test_score (decreasing)
candidates_sorted = candidates.sort_values(
    by=["max_depth", "min_samples_leaf", "mean_test_score"],
    ascending=[True, False, False]
).reset_index(drop=True)

# 4. Take the first one and create the 'chosen' dictionary with integer types
winning_candidate = candidates_sorted.iloc[0]

chosen = {
    "max_depth": int(winning_candidate["max_depth"]),
    "min_samples_leaf": int(winning_candidate["min_samples_leaf"]),
    "criterion": str(winning_candidate["criterion"])
}

print("\nChosen configuration (simplest within 1-SD noise):")
print(chosen)
```
